```python
import pandas as pd

df = pd.read_csv('uber_delivery_tips_dataset.csv')
```

```python
df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
```

```python
df['tip'] = df['tip'].fillna(0)
```

```python
df['earnings'] = df['earnings'].fillna(df['earnings'].mean())
```

```python
df = df.dropna()
```

```python
df = df.drop_duplicates()
```

```python
df.dtypes
```

```python
df['part_of_day'] = df['part_of_day'].str.strip()
```

```python
df = df[df['earnings'] > 0]
```

```python
df = df[df['earnings'] < 100]
```
