```python
import pandas as pd
```

```python
df = pd.read_csv('uber_delivery_tips_dataset.csv')
```

```python
df.head()
```

```python
df.groupby('part_of_day')['earnings'].mean()
```

```python
df.groupby('weekday_name')['earnings'].sum()
```

```python
df.groupby('part_of_day')['earnings'].agg(['mean', 'max', 'min'])
```

```python
df.groupby(['part_of_day', 'is_weekend'])['earnings'].mean()
```

```python
df.groupby('part_of_day').size()
```

```python
df.groupby('part_of_day')['tip'].mean()
```

```python
df.groupby('part_of_day')['earnings'].mean().sort_values(ascending=False)
```

```python
df.groupby('weekday_name')['earnings'].mean()
```

```python
df.groupby('is_weekend')['tip'].sum()
```

```python
df.groupby('part_of_day')['earnings'].mean().sort_values(ascending=False)
```
